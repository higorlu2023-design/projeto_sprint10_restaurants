```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
```

```python
df= pd.read_csv(r'C:\Users\iago_\projeto_10\data\rest_data_us_upd.csv')
df.head()
```

```python
df['address'].str.split().head()
#O método split  separa cada palavra e cria uma lista com estas palavras
```

```python
df['name_street'] = df['address'].str.split().str[1:].str.join(' ')
df['name_street'].head(10)
#utilizado o metodos split, fatiamento e join para criar a coluna com os nomes das ruas de cada estabelecimento. 
```

```python
df.head(10)
```

```python
tipos_estbl = df.groupby('object_type')['object_name'].count()
tipos_estbl.head(15)
```

```python
proporção_por_tipo = tipos_estbl / tipos_estbl.sum() * 100
```

```python
proporção_por_tipo.head(10)
```

```python
proporção_por_tipo = proporção_por_tipo.reset_index()
proporção_por_tipo.head()
```

```python
import plotly.express as px

fig = px.bar(
    proporção_por_tipo,
    x='object_type',
    y='object_name',
    color_discrete_sequence=['#2563EB'],
    width=900,
    height=500
)

fig.update_layout(
    title='Proporção por tipo de estabelecimento',
    xaxis_title='Tipo de estabelecimento',
    yaxis_title='Proporção (%)',
    template='plotly_white',
    showlegend=False
)

fig.show()
```

```python
df.head()
```

```python
pp_rede =  df.groupby('chain')['object_type'].count()
pp_rede.head(10)
```

```python
proporção_rede_indp = pp_rede / pp_rede.sum() * 100
proporção_rede_indp.head(10).round()
```

```python
proporção_rede_indp = proporção_rede_indp.reset_index()
```

```python
fig = px.bar(
    proporção_rede_indp,
    x='chain',
    y='object_type',
    color_discrete_sequence=['#2563EB'],
    width=900,
    height=500
)

fig.update_layout(
    title='Proporção de estabelecimentos por rede ou independente',
    xaxis_title='tipo de operação',
    yaxis_title='Proporção (%)',
    template='plotly_white',
    showlegend=True 
)

fig.show()
```
